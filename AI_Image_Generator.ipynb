```python
# Optional: connect this notebook to a GitHub repository.
# If this notebook is opened from the repository in Colab, set REPO_URL to that repository URL.
REPO_URL = "https://github.com/acherlab-ai/colab-ai-image"
BRANCH = "main"

import os, subprocess, sys
from pathlib import Path

WORKDIR = Path("/content/colab-ai-image-generator")
if WORKDIR.exists():
    subprocess.run(["bash", "-lc", f"rm -rf {WORKDIR}"], check=False)

# Replace YOUR_GITHUB_USERNAME after publishing, or upload the repository ZIP into Colab.
if "acherlab-ai/colab-ai-image" not in REPO_URL and "github.com/" not in REPO_URL:
    print("Set REPO_URL to a valid GitHub repository URL before running.")
    print("For a quick test, upload/extract the project files into /content/colab-ai-image-generator.")
else:
    subprocess.run(["git", "clone", "--depth", "1", "--branch", BRANCH, REPO_URL, str(WORKDIR)], check=True)

if not WORKDIR.exists():
    raise RuntimeError("Project files were not found. Set REPO_URL to your GitHub repo and rerun, or upload the project files.")
os.chdir(WORKDIR)
print("Project directory:", os.getcwd())
```
